```python
import torch

# Check if PyTorch can detect your GPU
is_available = torch.cuda.is_available()
print(f"Is CUDA available? {is_available}")

# If it's available, print the GPU's name
if is_available:
    gpu_name = torch.cuda.get_device_name(0)
    print(f"GPU Device: {gpu_name}")
```

```python
from ultralytics import YOLO

# Load a pretrained YOLOv8n model
model = YOLO('yolov8n.pt')

# Train the model using the powerful 3070 Ti
results = model.train(
   data="C:/Users/singh/Desktop/fish/nayadataset/data.yaml",       # Double-check this path
   epochs=100,
   batch=16,                          # Recommended batch size for 8GB VRAM
   imgsz=640,
   patience=20,
   device=0,                          # Use the GPU
   name='yolov8n_fish_detector_3070ti'
)
```

```python
from ultralytics import YOLO

# Load your trained model
model = YOLO(r"C:\Users\singh\Desktop\fish\nayadataset\runs\detect\yolov8n_fish_detector_3070ti\weights\best.pt")

# Run evaluation
metrics = model.val(
    data="C:/Users/singh/Desktop/fish/nayadataset/data.yaml",
    batch=16,
    imgsz=640,
    device=0,
    plots=True   # saves confusion matrix, PR curves, etc.
)

# Print summary metrics
print("Evaluation Metrics:")
print(f"Precision: {metrics.box.p.mean():.4f}")
print(f"Recall: {metrics.box.r.mean():.4f}")
print(f"mAP@0.5: {metrics.box.map50.mean():.4f}")
print(f"mAP@0.5:0.95: {metrics.box.map.mean():.4f}")
```
